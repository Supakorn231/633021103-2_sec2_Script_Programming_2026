# Activity 4.2 List vs. Tuple: Scenario Matching

| Scenario                                                              | Data Structure | Reason                                                                                                  |
| --------------------------------------------------------------------- | -------------- | ------------------------------------------------------------------------------------------------------- |
| 1. A collection of daily temperatures recorded over a week.           | **List**       | Temperatures may be updated or corrected, so a mutable structure is appropriate.                        |
| 2. The RGB color values for a specific shade (255, 0, 0).             | **Tuple**      | RGB values represent a fixed color and should not be changed, so an immutable structure is appropriate. |
| 3. Items in a shopping cart that can be added or removed.             | **List**       | Items can be added, removed, or modified, making a list the best choice.                                |
| 4. The days of the week in order.                                     | **Tuple**      | The days of the week are fixed and do not change, so a tuple is suitable.                               |
| 5. Student scores on a series of assignments, which might be updated. | **List**       | Scores may change after grading or corrections, so a mutable list is appropriate.                       |
| 6. GPS coordinates for a fixed landmark.                              | **Tuple**      | The coordinates of a fixed landmark do not change, so a tuple is the appropriate choice.                |


```python
# Activity 4.2 : Interactive To-Do List

todo_list = []

while True:
    print("\n===== TO-DO LIST MENU =====")
    print("1. Add a task")
    print("2. View tasks")
    print("3. Mark task as complete")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        task = input("Enter a new task: ")
        todo_list.append(task)
        print("Task added successfully.")

    elif choice == "2":
        if len(todo_list) == 0:
            print("No tasks in the to-do list.")
        else:
            print("\nYour Tasks:")
            for i in range(len(todo_list)):
                print(f"{i + 1}. {todo_list[i]}")

    elif choice == "3":
        if len(todo_list) == 0:
            print("No tasks to remove.")
        else:
            print("\nCurrent Tasks:")
            for i in range(len(todo_list)):
                print(f"{i + 1}. {todo_list[i]}")

            task_number = int(input("Enter the task number to complete: "))

            if 1 <= task_number <= len(todo_list):
                completed_task = todo_list.pop(task_number - 1)
                print(completed_task, "has been completed and removed.")
            else:
                print("Invalid task number.")

    elif choice == "4":
        print("Exiting To-Do List...")
        break

    else:
        print("Invalid choice. Please try again.")
```
