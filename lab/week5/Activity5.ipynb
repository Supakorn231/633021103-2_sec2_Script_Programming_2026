กิจกรรม 5.2: โปรแกรมนับความถี่ของคำ (Word Frequency Counter) โดยใช้ Dictionary

เขียนโปรแกรมตามขั้นตอนต่อไปนี้

รับข้อความ (Sentence) จากผู้ใช้
แปลงข้อความทั้งหมดให้เป็นตัวพิมพ์เล็กด้วย lower()
แยกข้อความออกเป็นคำด้วย split()
สร้าง Dictionary ว่างชื่อ word_counts
วนลูปทีละคำ
ตรวจสอบว่าคำนั้นมีอยู่ใน Dictionary หรือไม่
ถ้ามีแล้ว ให้เพิ่มจำนวนครั้ง (+1)
ถ้ายังไม่มี ให้เพิ่มคำนั้นเข้าไปพร้อมกำหนดค่าเริ่มต้นเป็น 1
แสดงผล Dictionary ที่เก็บจำนวนครั้งของแต่ละคำ


# ขั้นตอนที่ 1: รับข้อความจากผู้ใช้
sentence = input("กรุณาป้อนข้อความ: ")

# ขั้นตอนที่ 2: แปลงเป็นตัวพิมพ์เล็ก
sentence = sentence.lower()

# ขั้นตอนที่ 3: แยกข้อความออกเป็นคำ
words = sentence.split()

# ขั้นตอนที่ 4: สร้าง Dictionary ว่าง
word_counts = {}

# ขั้นตอนที่ 5-6: นับจำนวนคำ
for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# ขั้นตอนที่ 7: แสดงผลลัพธ์
print("\nจำนวนครั้งของแต่ละคำ")
print(word_counts)


```python
ตัวอย่างโค้ด (Challenge)
# ขั้นตอนที่ 1: รับข้อความ
sentence = input("กรุณาป้อนข้อความ: ")

# ขั้นตอนที่ 2: แปลงเป็นตัวพิมพ์เล็ก
sentence = sentence.lower()

# ขั้นตอนที่ 3: แยกข้อความเป็นคำ
words = sentence.split()

# ขั้นตอนที่ 4: สร้าง Dictionary ว่าง
word_counts = {}

# ขั้นตอนที่ 5-6: ลบเครื่องหมายวรรคตอนและนับคำ
for word in words:
    word = word.strip('.,!?"()')

    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# ขั้นตอนที่ 7: แสดงผล
print("\nจำนวนครั้งของแต่ละคำ")
print(word_counts)
```
