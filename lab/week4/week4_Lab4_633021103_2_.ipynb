# lab4_1.py - Student Management System (List Operations)

```python
# Lab 4.1 : Student Management System (List Operations)

student_names = []

# -----------------------------
# 1-7 Basic Tasks
# -----------------------------
print("=== Add 3 Student Names ===")

for i in range(3):
    name = input(f"Enter student name {i + 1}: ")
    student_names.append(name)

print("\nCurrent Student List:")
print(student_names)

# Find a student
search_name = input("\nEnter a student name to search: ")

if search_name in student_names:
    print(search_name, "found at index", student_names.index(search_name))
else:
    print(search_name, "not found.")

# Remove a student
remove_name = input("\nEnter a student name to remove: ")

if remove_name in student_names:
    student_names.remove(remove_name)
    print(remove_name, "has been removed.")
else:
    print(remove_name, "not found.")

print("Updated Student List:")
print(student_names)

# Sort students
student_names.sort()

print("\nSorted Student List:")
print(student_names)

# Count students
print("Total number of students:", len(student_names))

# -----------------------------
# 8 Challenge
# -----------------------------
while True:
    print("\n===== Student Management Menu =====")
    print("1. Add Student")
    print("2. Remove Student")
    print("3. Search Student")
    print("4. Sort Students")
    print("5. Display Students")
    print("6. Count Students")
    print("7. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter student name: ")
        student_names.append(name)
        print("Student added.")

    elif choice == "2":
        name = input("Enter student name to remove: ")

        if name in student_names:
            student_names.remove(name)
            print("Student removed.")
        else:
            print("Student not found.")

    elif choice == "3":
        name = input("Enter student name to search: ")

        if name in student_names:
            print(name, "found at index", student_names.index(name))
        else:
            print("Student not found.")

    elif choice == "4":
        student_names.sort()
        print("Student list sorted.")
        print(student_names)

    elif choice == "5":
        print("Student List:")
        print(student_names)

    elif choice == "6":
        print("Total students:", len(student_names))

    elif choice == "7":
        print("Exiting program...")
        break

    else:
        print("Invalid choice. Please try again.")''
```

# lab4_2.py - Geometric Calculations (Tuples & Immutability)

```python
# Lab 4.2 : Geometric Calculations (Tuples & Immutability)

import math

# 1-2. Define Points
point1 = (3, 4)
point2 = (6, 8)

# 3. Access Coordinates
print("Point 1: x =", point1[0], ", y =", point1[1])
print("Point 2: x =", point2[0], ", y =", point2[1])

# 4. Attempt Modification (Tuple is immutable)
# Uncomment the line below to see the TypeError
# point1[0] = 5

print("\nTuples are immutable, so their values cannot be changed.")
print("Trying to assign a new value such as point1[0] = 5")
print("will cause: TypeError: 'tuple' object does not support item assignment")

# 5. Calculate Distance
x1 = point1[0]
y1 = point1[1]
x2 = point2[0]
y2 = point2[1]

distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

print("\nDistance between Point 1 and Point 2 =", distance)

# 6. Challenge
print("\n=== Calculate Distance Between Two New Points ===")

x1 = float(input("Enter x-coordinate of Point 1: "))
y1 = float(input("Enter y-coordinate of Point 1: "))
x2 = float(input("Enter x-coordinate of Point 2: "))
y2 = float(input("Enter y-coordinate of Point 2: "))

new_point1 = (x1, y1)
new_point2 = (x2, y2)

distance = math.sqrt((new_point2[0] - new_point1[0]) ** 2 +
                     (new_point2[1] - new_point1[1]) ** 2)

print("Point 1:", new_point1)
print("Point 2:", new_point2)
print("Distance =", distance)
```
